```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(-10, 10, 0.1)
```

```python
def plot(x, y):
    plt.scatter(x, y)
    plt.show()
```

### Step / Binary

```python
def step(z):
    if z < 0:
        return 0
    else:
        return 1
    # or
    # np.where(z >= 0, 1, 0)
```

```python
plot(x, [step(i) for i in x])
```

### Sign

```python
def sign(z):
    if z < 0:
        return -1
    elif z == 0:
        return 0
    else:
        return 1

    # or
    # np.sign(z)
```

```python
plot(x, [sign(i) for i in x])
```

### Softplus

```python
def softplus(z):
    return np.log(1 + np.exp(z))
```

```python
plot(x, softplus(x))
```

### ReLU

```python
def relu(z):
    return np.maximum(0, z)
```

```python
plot(x, relu(x))
```

### Leaky ReLU

```python
def leaky_relu(z, alpha=0.01):
    return np.maximum(alpha*z, z)
```

```python
plot(x, leaky_relu(x, 0)) # alpha = 0 : ReLU
```

```python
plot(x, leaky_relu(x, 0.01))
```

```python
plot(x, leaky_relu(x, 0.1))
```

```python
plot(x, leaky_relu(x, 0.5))
```

```python
plot(x, leaky_relu(x, 1)) # alpha = 1 : linear
```

### PReLU

PRelu is Leaky Relu where alpha is Learned by Model

### RReLU

```python
def random_relu(z, lower=1/8, upper=1/3):
    alpha = np.random.uniform(lower, upper, size=z.shape)
    return np.maximum(alpha*z, z)
```

```python
plot(x, random_relu(x, 0.1, 0.1))
```

```python
plot(x, random_relu(x, 1/8, 1/3))
```

```python
plot(x, random_relu(x, 0, 1))
```

### ELU

```python
def elu(z, alpha=1):
    if z < 0:
        return alpha * (np.e**z - 1)
    else:
        return z
    # or
    # np.where(z > 0, z, alpha * (np.exp(z) - 1))
```

```python
plot(x, [elu(i, alpha=0) for i in x])
```

```python
plot(x, [elu(i, alpha=0.1) for i in x])
```

```python
plot(x, [elu(i, alpha=0.5) for i in x])
```

```python
plot(x, [elu(i, alpha=1) for i in x])
```

```python
plot(x, [elu(i, alpha=2) for i in x])
```

```python
plot(x, [elu(i, alpha=5) for i in x])
```

### SELU

```python
def selu(z):
    alpha = 1.67326324
    scale = 1.05070099
    if z < 0:
        return scale * elu(z, alpha)
    else:
        return z
```

```python
plot(x, [selu(i) for i in x])
```

### GELU

```python
from scipy.special import erf

def gelu(z):
    return 0.5 * z * (1 + erf(z / np.sqrt(2)))
```

```python
plot(x, gelu(x))
```

### Swish / SiLU

```python
def swish(z):
    return z / (1 + np.exp(-z))
```

```python
plot(x, swish(x))
```

### Mish

```python
def mish(z):
    softplus = np.log(1 + np.exp(z))
    return x * np.tanh(softplus)
```

```python
plot(x, mish(x))
```
