fan-in = Number of Input Connection to the current neuron

fan-out = Number of Output Collection to the current neuron

fan-avg = (fan-in + fan-out) / 2


Normal Initialization

W ~ N(0, Var(W)) # Random Sample Weights from Normal Distribution with Mean 0, Variance Var(W)

Var(W) = 2 / fan-in = 1 / (fan-in/2) # ReLU reduces input by Half on average

Uniform Initialization

W ~ U(-r, +r) # Random Sample Weights from Uniform Distribution in range -r to +r

r = sqrt(6 / fan-in) = sqrt(3 / (fan-in/2))

### Simple Code

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def normal_init(fan_in):
    var = 2/fan_in
    return np.random.normal(0, var**0.5) # Numpy takes standard deviation, sd = var^0.5
```

```python
def uniform_init(fan_in):
    r = (6 / fan_in)**0.5
    return np.random.uniform(-r, r)
```

```python
fan_in = 1000
n = 100
x = []
y = []
print("Variance :", 2/fan_in)
for i in range(n):
    x.append(i)
    w = normal_init(fan_in)
    y.append(w)

plt.scatter(x, y)
plt.show()
```

```python
fan_in = 1000
n = 100
x = []
y = []
print("Range r :", (6 / fan_in)**0.5)
for i in range(n):
    x.append(i)
    w = uniform_init(fan_in)
    y.append(w)

plt.scatter(x, y)
plt.show()
```

### In Torch

```python
import torch
import torch.nn as nn
```

#### Uniform Init

```python
layer = nn.Linear(40, 10) # # nn.Linear module initializes its weights using Kaiming uniform initialization
nn.init.kaiming_uniform_(layer.weight)
nn.init.zeros_(layer.bias)
```

#### Normal Init

```python
layer = nn.Linear(40, 10) # # nn.Linear module initializes its weights using Kaiming uniform initialization
nn.init.kaiming_normal_(layer.weight)
nn.init.zeros_(layer.bias)
```
