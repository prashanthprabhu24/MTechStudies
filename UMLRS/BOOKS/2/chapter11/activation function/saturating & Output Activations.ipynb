```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(-10, 10, 0.1)
```

```python
def plot(x, y):
    plt.scatter(x, y)
    plt.show()
```

### Sigmoid

```python
def sigmoid(z):
    return 1 / (1+ np.e**(-z))
```

```python
plot(x, sigmoid(x))
```

### Tanh

```python
def tanh(z):
    return (np.e**x - np.e**(-x))/(np.e**x + np.e**(-x))
```

```python
plot(x, tanh(x))
```

### Hard Sigmoid

```python
def hard_sigmoid(z):
    if z <= -2.5:
        return 0
    elif z <= 2.5:
        return 0.2*z + 0.5
    else:
        return 1

    # Or (vectorized)
    # return np.clip(0.2 * z + 0.5, 0.0, 1.0)
```

```python
plot(x, [hard_sigmoid(i) for i in x])
```

### Hard Tanh

```python
def hard_tanh(z):
    if z <= -1:
        return -1
    elif z <= 1:
        return z
    else:
        return 1

    # Or (vectorized)
    # return np.clip(z, -1.0, 1.0)
```

```python
plot(x, [hard_tanh(i) for i in x])
```

### Softmax

```python
def softmax(z):
    return np.exp(z) / np.sum(np.exp(z))
```

```python
z = np.array([10.0, 2.0, 1.0, 0.1, 0, -10])
result = softmax(z)
result, sum(result)
```

### LogSoftmax

```python
def log_softmax(z):
    return z - np.log(np.sum(np.exp(z)))
```

```python
z = np.array([10.0, 2.0, 1.0, 0.1, 0, -10])
result = log_softmax(z)
result, sum(result)
```
