```python
import torch.nn as nn
```

```python
model = nn.Sequential(nn.Flatten(), nn.BatchNorm1d(1 * 28 * 28),
                      nn.Linear(1 * 28 * 28, 300), nn.ReLU(), nn.BatchNorm1d(300),
                      nn.Linear(300, 100), nn.ReLU(), nn.BatchNorm1d(100),
                      nn.Linear(100, 10))
```
