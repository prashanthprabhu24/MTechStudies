```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import Ridge, Lasso, ElasticNet
from sklearn.linear_model import SGDRegressor
```

### Load data

```python
rng = np.random.default_rng(seed=42)
m = 50
X = rng.random((m, 1))
X
```

```python
m_ = 3
c_ = 4
noise = rng.standard_normal((m, 1))
y = c_ + m_ * X + noise
y
```

```python
plt.scatter(X, y)
plt.grid()
plt.show()
```

### Rigid

```python
# Ridge regression closed-form solution
ridge_reg_0 = Ridge(alpha=0, solver="cholesky")
ridge_reg_0.fit(X, y)
ridge_reg_0.intercept_, ridge_reg_0.coef_
```

```python
# Ridge regression closed-form solution
ridge_reg_01 = Ridge(alpha=0.1, solver="cholesky")
ridge_reg_01.fit(X, y)
ridge_reg_01.intercept_, ridge_reg_01.coef_
```

```python
# Ridge regression closed-form solution
ridge_reg_1 = Ridge(alpha=1, solver="cholesky")
ridge_reg_1.fit(X, y)
ridge_reg_1.intercept_, ridge_reg_1.coef_
```

```python
# Ridge regression closed-form solution
ridge_reg_10 = Ridge(alpha=10, solver="cholesky")
ridge_reg_10.fit(X, y)
ridge_reg_10.intercept_, ridge_reg_10.coef_
```

```python
# Ridge regression closed-form solution
ridge_reg_100 = Ridge(alpha=100, solver="cholesky")
ridge_reg_100.fit(X, y)
ridge_reg_100.intercept_, ridge_reg_100.coef_
```

```python
plt.scatter(X, y)
plt.plot(X, ridge_reg_0.intercept_[0] + ridge_reg_0.coef_[0]*X, label="alpha = 0")
#plt.plot(X, ridge_reg_01.intercept_[0] + ridge_reg_01.coef_[0]*X, label="alpha = 0.1")
plt.plot(X, ridge_reg_1.intercept_[0] + ridge_reg_1.coef_[0]*X, label="alpha = 1")
plt.plot(X, ridge_reg_10.intercept_[0] + ridge_reg_10.coef_[0]*X, label="alpha = 10")
plt.plot(X, ridge_reg_100.intercept_[0] + ridge_reg_100.coef_[0]*X, label="alpha = 100")

plt.legend()
plt.grid()
plt.show()
```

```python
# Ridge regression SGD
sgd_reg_0 = SGDRegressor(penalty="l2", alpha=0 / m, tol=None, max_iter=1000, eta0=0.01, random_state=42)
sgd_reg_0.fit(X, y.ravel())
sgd_reg_0.intercept_, sgd_reg_0.coef_
```

```python
# Ridge regression SGD
sgd_reg_01 = SGDRegressor(penalty="l2", alpha=0.1 / m, tol=None, max_iter=1000, eta0=0.01, random_state=42)
sgd_reg_01.fit(X, y.ravel())
sgd_reg_01.intercept_, sgd_reg_01.coef_
```

```python
# Ridge regression SGD
sgd_reg_1 = SGDRegressor(penalty="l2", alpha=1 / m, tol=None, max_iter=1000, eta0=0.01, random_state=42)
sgd_reg_1.fit(X, y.ravel())
sgd_reg_1.intercept_, sgd_reg_1.coef_
```

```python
# Ridge regression SGD
sgd_reg_10 = SGDRegressor(penalty="l2", alpha=10 / m, tol=None, max_iter=1000, eta0=0.01, random_state=42)
sgd_reg_10.fit(X, y.ravel())
sgd_reg_10.intercept_, sgd_reg_10.coef_
```

```python
# Ridge regression SGD
sgd_reg_100 = SGDRegressor(penalty="l2", alpha=100 / m, tol=None, max_iter=1000, eta0=0.01, random_state=42)
sgd_reg_100.fit(X, y.ravel())
sgd_reg_100.intercept_, sgd_reg_100.coef_
```

```python
plt.scatter(X, y)
plt.plot(X, sgd_reg_0.intercept_[0] + sgd_reg_0.coef_[0]*X, label="alpha = 0")
#plt.plot(X, sgd_reg_01.intercept_[0] + sgd_reg_01.coef_[0]*X, label="alpha = 0.1")
plt.plot(X, sgd_reg_1.intercept_[0] + sgd_reg_1.coef_[0]*X, label="alpha = 1")
plt.plot(X, sgd_reg_10.intercept_[0] + sgd_reg_10.coef_[0]*X, label="alpha = 10")
plt.plot(X, sgd_reg_100.intercept_[0] + sgd_reg_100.coef_[0]*X, label="alpha = 100")

plt.legend()
plt.grid()
plt.show()
```
